```python
import numpy as np
import pandas as pd
from pred import predict_all
```

```python
# read each of the csv files as a *pandas data frame*
dataset = pd.read_csv("cleaned_data_combined.csv",
                      keep_default_na=False         # Prevent pandas from auto-converting "None"-like strings
                      )
dataset.columns = ["ID", "P1", "P2", "P3", "P4", "P5", "P6", "P7", "P8", "Label"]

dataset_no_label = dataset.drop(columns=["Label"])
dataset_no_label.to_csv("cleaned_data_combined_no_label.csv", index=False)
```

```python
# Test predict_all function on validation dataset.
output = predict_all("cleaned_data_combined_no_label.csv")
correct = 0
for i in range(len(output)):
    if output[i] == dataset["Label"][i]:
        correct += 1
print("Accuracy: " + str(correct / len(output)))
```
