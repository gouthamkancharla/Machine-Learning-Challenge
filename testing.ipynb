```python
from pred import predict_all
import pandas as pd
df_test = pd.read_csv("cleaned_data_combined.csv", keep_default_na=False)
import random
```

```python
output = predict_all("cleaned_data_combined.csv")
accuracy_list = []
for j in range(0, 1000):
    x = random.randint(0, len(df_test)-round(len(df_test)*0.2)-1) # Chooses a random index to start at
    count = 0
    for i in range(round(len(df_test)*0.2)):
        if df_test["Label"][x+i] == output[x+i]:
            count += 1
    print("Accurcy: " + str(count / round(len(df_test)*0.2)))
    accuracy_list.append(count / round(len(df_test)*0.2))
```

```python
print("average acc: " + str((sum(accuracy_list) / len(accuracy_list))))
```
